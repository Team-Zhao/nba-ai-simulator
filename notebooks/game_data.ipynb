```python
from nba_api.stats.endpoints import leaguegamelog, boxscoretraditionalv3
```

```python
game_log = leaguegamelog.LeagueGameLog(
    season="2025-26",
    season_type_all_star="Regular Season",
)

df = game_log.get_data_frames()[0]
```

```python
df.head()
```

```python
df[[
    "GAME_ID",
    "GAME_DATE",
    "TEAM_ABBREVIATION",
    "MATCHUP",
    "WL",
    "PTS",
    "PLUS_MINUS",
]].head(10)
```

```python
game_id = df.iloc[0]["GAME_ID"]
game_id
```

```python
box = boxscoretraditionalv3.BoxScoreTraditionalV3(
    game_id=game_id
)
box_dfs = box.get_data_frames()
len(box_dfs)
```

```python
player_df = box_dfs[0]

player_df.head()
```
